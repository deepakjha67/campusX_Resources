### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Write your code here
import math
x1 = int(input("Enter the value of x1: "))
y1 = int(input("Enter the value of y1: "))
x2 = int(input("Enter the value of x2: "))
y2 = int(input("Enter the value of y2: "))

distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

print ("Euclidian distance: ", distance)
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here
heads = int(input("Enter the total heads: "))
legs = int(input("Enter the total legs: "))
dogs = (legs - 2 * heads) // 2
chickens = heads - dogs

if(legs % 2 != 0 or dogs < 0 or chickens < 0 or 
   (dogs + chickens != heads) or 
   (4 * dogs + 2 * chickens != legs)):
    print("Invalid Input")
else:
    print("Dogs =", dogs)
    print("chickens = ", chickens)

````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
# Write your code here
n = int(input("Enter the value of n: "))
sum_of_squares = 0
for i in range(1, n+1):
    sum_of_squares += i ** 2

print("Sum of first n natural number is: ", sum_of_squares)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here
a = int(input("Enter the first term: "))
b = int(input("Enter the second term: "))
n = int(input("Enter the value of n: "))

d = b - a
nth_term = a + (n - 1) * d
print("The", n, "th term is: ", nth_term)
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here
import math

# Dimensions of the milk tank
H = float(input("Enter height of tank (cm): "))
L = float(input("Enter length of tank (cm): "))
B = float(input("Enter breadth of tank (cm): "))

# Dimensions of the glass
h = float(input("Enter height of glass (cm): "))
r = float(input("Enter radius of glass (cm): "))

# Volume of the milk tank
tank_volume = H * L * B

# Volume of one glass
glass_volume = math.pi * r * r * h

# Number of glasses
glasses = tank_volume / glass_volume

print("Number of glasses that can be filled:", int(glasses))
````
